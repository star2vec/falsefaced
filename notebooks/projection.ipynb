# projection

**Question.** What do the layer-12 activations look like in two dimensions, per concept, under no trigger, the own-concept trigger and an other-concept trigger?

For viewing only: 2 of 3584 dimensions. Nothing is measured here.

```python
# cell 1: imports and data, run once
import sys; sys.path.insert(0, '..')
import numpy as np, pandas as pd
from lib import SEED, plots
from lib.paths import ACTS, META

acts = np.load(ACTS).astype(np.float32)
meta = pd.read_csv(META)
acts.shape, len(meta)
```

## Write-up figure 7

Per concept: PCA to 2 dimensions fitted on that concept's rows (all three conditions, all prompts), points coloured by condition. Saved as `figures/7_projection.png` and `.svg`.

```python
plots.style()
fig = plots.projection(acts, meta)
plots.save(fig, "7_projection")
```

## What the plot shows

For every concept the three conditions form three separate clusters in the first two principal components. The other-concept trigger moves the activations about as far from the no-trigger rows as the own-concept trigger does, in a different direction. So the largest directions of variation within a concept are set by which trigger is in the prompt, not only by hiding. A picture in 2 of 3584 dimensions, for viewing only; no number is taken from it.
